# Applied Domino Workshop: start here
Run this notebook once, top to bottom, in a JupyterLab workspace in your new project. It downloads the workshop code into the project and puts the raw data into the project's Domino Dataset.

If your repository is private or internal, set a GitHub token first (see the next cell).

```python
import os

# Where the files are downloaded from. Change this only if your instructor gave you a different address.
RAW_BASE = os.environ.get("WORKSHOP_RAW_BASE", "https://raw.githubusercontent.com/DDL-Mike-Snyder/fraud-workshop/main/resources")

# Only needed if the GitHub repository is not public. Create a token that can read the repository,
# then run:  os.environ["GITHUB_TOKEN"] = "ghp_..."
TOKEN = os.environ.get("GITHUB_TOKEN", "")
AUTH = f'-H "Authorization: token {TOKEN}"' if TOKEN else ""
print("Downloading from", RAW_BASE)
```

## 1. Download the code and the data

```python
!curl -fSL {AUTH} -o fraud-workshop-code.zip {RAW_BASE}/fraud-workshop-code.zip
!curl -fSL {AUTH} -o fraud-workshop-data.zip {RAW_BASE}/fraud-workshop-data.zip
!ls -lh fraud-workshop-*.zip
```

## 2. Unzip the code into the project root

```python
import zipfile
with zipfile.ZipFile("fraud-workshop-code.zip") as z:
    z.extractall(".")
print("exercises/ present:", os.path.isdir("exercises"))
```

## 3. Put the raw data in the project's Domino Dataset

```python
username = os.environ.get("DOMINO_USER_NAME")
project = os.environ.get("DOMINO_PROJECT_NAME", "my-local-project")
datasets = os.environ.get("DOMINO_DATASETS_DIR", "/domino/datasets/")
if os.environ.get("DOMINO_IS_GIT_BASED") == "true":
    dataset_dir = f"{datasets}/{project}/{username}"
else:
    dataset_dir = f"{datasets}/local/{project}"
os.makedirs(dataset_dir, exist_ok=True)
with zipfile.ZipFile("fraud-workshop-data.zip") as z:
    z.extractall(dataset_dir)
print("Dataset folder:", dataset_dir)
print(os.listdir(dataset_dir))
assert os.path.exists(f"{dataset_dir}/raw_cc_transactions.csv.gz"), "raw_cc_transactions.csv.gz is missing"
```
